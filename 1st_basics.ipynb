```python
import numpy as np
```

```python
a = [2,5,7,8,9]
print(np.array(a))
```

```python
a = np.array([10,39,59,50,60])
b = np.array([30,50,46,68,89])
result = np.add(a,b)
print(result)
```

```python
a = np.array([10,39,59,50,60])
b = np.array([30,50,46,68,89])
result = np.subtract(a,b)
print(result)
```

```python
a = np.array([10,39,59,50,60])
b = np.array([30,50,46,68,89])
result = np.multiply(a,b)
print(result)
```

```python
a = np.array([10,39,59,50,60])
b = np.array([30,50,46,68,89])
result = np.divide(a,b)
print(result)
```

```python
a = np.array([10,39,59,50,60])
b = np.array([3,2,2,2,2])
result = np.power(a,b)
print(result)
```

```python
a = np.array([10,39,59,50,60])
b = np.array([3,5,4,6,8])
result = np.mod(a,b)
print(result)
```
